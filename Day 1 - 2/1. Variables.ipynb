# Veri Türleri (Data Types)
Python'da temel veri türleri şunlardır:

## 1. Sayısal Veri Türleri
- **int**: Tam sayılar (1, -5, 1000)
- **float**: Ondalıklı sayılar (3.14, -0.001, 2.0)
- **complex**: Karmaşık sayılar (3+4j)

## 2. Metin Veri Türü
- **str**: Metinsel ifadeler ("Merhaba", 'Python')

## 3. Mantıksal Veri Türü
- **bool**: True veya False değerlerini alır

## 4. Dizi Veri Türleri
- **list**: Sıralı ve değiştirilebilir koleksiyon [1, 2, 3]
- **tuple**: Sıralı ve değiştirilemez koleksiyon (1, 2, 3)
- **set**: Sırasız ve benzersiz elemanlar {1, 2, 3}
- **dict**: Anahtar-değer çiftleri {"anahtar": "değer"}

## type() Fonksiyonu
- Bir değişkenin veya ifadenin veri türünü öğrenmek için kullanılır
- Örnek:
  ```python
  x = 5
  print(type(x))  # <class 'int'>
  
  y = "Merhaba"
  print(type(y))  # <class 'str'>
  ```

# Değişken Adlandırma Kuralları ve Şekilleri

Temel Kurallar:
- Değişken isimleri harf veya alt çizgi (_) ile başlamalıdır
- Rakam ile başlayamaz ama içerebilir
- Boşluk içeremez
- Python'ın özel kelimeleri (if, for, while vb.) kullanılamaz
- Büyük-küçük harf duyarlıdır (age ve Age farklı değişkenlerdir)

Yaygın Adlandırma Şekilleri:

1. Camel Case (camelCase)
- İlk kelime küçük, sonraki kelimelerin ilk harfi büyük yazılır
- Örnek:
  <br>
  firstName = "Ahmet"
  <br>
  lastName = "Yılmaz"
  <br>
  dateOfBirth = "2000"

2. Snake Case (snake_case)
- Kelimeler alt çizgi (_) ile birbirine bağlanır
- Python'da en çok tercih edilen yöntemdir
- Örnek:
  <br>
  first_name = "Ahmet"
  <br>
  last_name = "Yılmaz"
    <br>
  date_of_birth = "2000"
