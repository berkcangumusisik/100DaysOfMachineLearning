- \n karakteri string içerisinde yeni satır oluşturmak için kullanılır. "new line" anlamına gelir.


- \t karakteri string içerisinde tab boşluğu oluşturmak için kullanılır. "tab" anlamına gelir.

```python
print("Hello\nWorld")
```

```python
print("Hello\tWorld")
```
